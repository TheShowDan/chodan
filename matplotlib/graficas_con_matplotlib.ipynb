### Generación de datos

#### Para "x"

Antes que nada, se tienen que generar los datos que se van a graficar, vamos a a ir primero por una función de una sola varíable, por lo que, se necesitaran los datos de "x", y de "y". se comenzará con la varíable independiente "x", y a continuación se proponen dos formas de generar esos datos, con ayuda de NumPy.

La primer forma es con un np.arrange(), este método tiene tres argumentos, que a groso modo es: Dime de donde a donde quieres que vaya y de que tamaño van a ser los intervalos con los que voy a avanzar.

- np.arange(inicio, final, tamaño de intervalo)

```python
x0 = np.arange(0,26,2)
x0
```

Otra opción es generar los datos de "x" con np.linspace(), que tambien tiene tres argumentos, sin embargo, este método los distribuye diferente, sería como: Dime de donde a donde quieres que vaya y cuantos pasos quieres que de para completar ese camino.

- np.linspace(inicio, final, cantidad de diviciones del intervalo)

```python
x = np.linspace(0,25,100)
x
```

### Graficación

A continuacion se muestra un ejemplo de las propiedades básicas de una gráfica

```python
#Datos
x = np.linspace(0,25,100)
y = np.sin(x)
y2 = np.cos(x)

#Esta linea siempre debe ir primero para modificar el tamaño del gráfico
plt.figure(figsize=(7,2))

#Funciones (lw = line weight | ms = marker size (tamaño bolitas))
plt.plot(x,y,'o',label='f(x)=sen(x)',color='steelblue',lw=0.7,ms=2)
plt.plot(x,y2,label='f(x)=cos(x)',color='indigo',lw=0.7)
plt.legend(loc='lower right',fontsize=8,ncol=2,edgecolor='darkgray')

#Titulos y Nombres ejes
plt.title('Ondas',fontsize=12)
plt.xlabel('Tiempo (s)',fontsize=10)
plt.ylabel('Rango (m)',fontsize=10)

#Limites de los ejes
plt.xlim(0,25)
plt.ylim(-1.75,1.5)

#Tamaño de los números de los ejes
plt.tick_params(axis='both',labelsize=9.5)

#Mostrar
plt.grid(True)
plt.show()
```

### Lo mismo pero separado (subplots)

Imagina que las graficas te las piden por separado, que quieren que en una sola figua esté el seno y el coseno pero en graficas aparte, para esto nos ayudaremos de los subplots

- subplots(cantidad de graficos en x, cantidad de graficos en y, número del gráfico actual)}

Un axe es un cuadrito individual

La figura (fig) es la imagen en conjunto, son todos los axes.

```python
#Datos
x = np.linspace(0,25,50)
y = np.sin(x)
y2 = np.cos(x)
y3 = x
y4 = x**(1/2)

#Se definen cuantos gráficos serán y de que tamaño 
fig, axes = plt.subplots(2,2,figsize=(6,6))

#Ahora se programa cada gráfico, parecido a lo que se hizo arriba pero individual
#Primero
ax = axes[0][0]
ax.plot(x,y,'--',color='navy',lw=0.7)
ax.set_xlabel('Tiempo (s)',fontsize=8)
ax.set_ylabel('$sen(x)$',fontsize=8)
ax.tick_params(axis='both',labelsize=8)
ax.grid(True)

#Segundo
ax = axes[0][1]
ax.plot(x, y2, 'o--', color='darkred', lw=0.7,ms=3)
ax.set_xlabel('$Tiempo [s]$', fontsize=9)
ax.set_ylabel('$cos(x)$', fontsize=9)
ax.tick_params(axis='both', labelsize=8)
ax.grid(True)

#Tercero
ax = axes[1][0]
ax.plot(x, y3, '--',color='deeppink', lw=0.7, label='Simulación',ms=2)
ax.set_xlabel('Tiempo [s]', fontsize=9)
ax.set_ylabel('$x$', fontsize=9)
ax.tick_params(axis='both', labelsize=8)
ax.legend(loc='upper left', fontsize=8, edgecolor='black')
ax.grid(True)

#Cuarto
ax = axes[1][1]
ax.plot(x, y4, 'o',color='forestgreen', lw=0.7, ms=2)
ax.set_xlabel('$Tiempo [s]$', fontsize=9)
ax.set_ylabel('$\sqrt{x}$', fontsize=9)
ax.tick_params(axis='both', labelsize=8)
ax.grid(True)

#Se ajustan los parametros de los subplots para que no estén amontonados, los valores de top, bot, left y right son de aquí son los de default
#Cambia los de hspace y wspace pasa separar
plt.subplots_adjust(hspace=0.4, wspace=0.45)
plt.show()
```

### Datos Experimentales

Con lo aprendido arriba se puede hacer una buena grafica con datos experimentales, lo cual incluye las barras de error.

Se empieza con los datos a graficar

errorbar es una herramienta que nos ayuda a graficar contemplando ya las barras de errores, algunos de los argumentos son:

- fmt - Tipo de linea o marca
- capthick - Indica el ancho de las líneas en los extremos (caps)
- capsize - Tamaño del cap
- elinewidth - Grosor de la linea de error

```python
#Datos Muestra 1
x1 = np.array([1.98, 3.02, 5.1, 7.64, 9.72, 11.28]) #pH
y1 = np.array([20.27, 14.08, 21.8, -54.56, -103.9, -119.9]) #mV
dest1 = np.array([4.57, 1.99, 3.39, 7.46, 7.81, 9.69])

#Datos Muestra 2
x2 = np.array([1.66, 2.96, 4.8, 7.21, 10.37, 11.54]) #pH
y2 = np.array([0, -16.39, -31.51, -53.43, -78.97, -94.39]) #mV
dest2 = np.array([0, 7.3, 6.43, 6.51, 6.91, 8.19])

#Datos Muestra 3
x3 = np.array([1.61, 3.05, 5.01, 6.43, 10.26, 11.65]) #pH
y3 = np.array([35.49, -23.33, -72.36, -87.77, -112.9, -102.31]) #mV
dest3 = np.array([4.7, 5.25, 5.72, 13.37, 14.47, 12.53])

#Se crea una figura y un solo subplot
fig, ax = plt.subplots(figsize=(10,5))

#Graficar datos
ax.errorbar(x1,y1,dest1, fmt='o--', capsize=6, elinewidth=0.75, capthick=0.75, color='indigo', lw=0.5, ms=5, label='Muestra 1')
ax.errorbar(x2,y2,dest1, fmt='o--', capsize=6, elinewidth=0.75, capthick=0.75, color='deeppink', lw=0.5, ms=5, label='Muestra 2')
ax.errorbar(x3,y3,dest1, fmt='o--', capsize=6, elinewidth=0.75, capthick=0.75, color='steelblue', lw=0.5, ms=5, label='Muestra 3')
plt.legend(loc='upper right', fontsize=9, edgecolor='darkgray')

#Titulos y Nombres ejes
ax.set_xlabel('pH',fontsize=12)
ax.set_ylabel('Potencial Zeta (mV)',fontsize=12)
ax.tick_params(axis='both',labelsize=10)

#Limite de Ejes
plt.xlim(1,12)

#Mostrar
plt.grid(True)
plt.show()
```

### Graficos de Barras

```python
#Datos
categorias = ['A', 'B', 'C', 'D', 'E']
valores = [5, 7, 3, 8, 4]
colores = ['steelblue','slategrey','teal','silver','slateblue']

#Total y porcentajes
total = sum(valores)
porcentajes = [valor/total * 100 for valor in valores]

#Grafico
plt.figure(figsize=(5,4))
plt.bar(categorias, valores,color=colores)

#Ajustes de Ejes
plt.xlabel('Categorias',fontsize=10)
plt.ylabel('Valores',fontsize=10)
plt.tick_params(axis='both',labelsize=8.5)
plt.ylim(0,9)

#Valores arriba de las barras
for i, (valor, porcentajes) in enumerate(zip(valores, porcentajes)):
    plt.text(i, valor + 0.1, f'{valor}\n{porcentajes:.1f}%',
    ha='center',fontsize=8)
    

#Mostrar
plt.show()
```

```python
#Datos
categorias = ['A', 'B', 'C', 'D', 'E']
valores = [5, 7, 3, 8, 4]
colores = ['steelblue','slategrey','dimgrey','darkcyan','slateblue']
err = [1.2, 0.5, 0.8, 0.4, 0.15]

#Total y porcentajes
total = sum(valores)
porcentajes = [valor / total * 100 for valor in valores]

#Grafico
plt.figure(figsize=(5,4))
plt.barh(categorias, valores, color='none',edgecolor=colores,linewidth=1,
        xerr=err, capsize=8, ecolor='#708090')

# Añadir títulos y etiquetas
plt.ylabel('Categorias',fontsize=10)
plt.xlabel('Valores',fontsize=10)
plt.tick_params(axis='both',labelsize=8.5)
plt.xlim(0,9)

# Mostrar el gráfico
plt.show()
```

### Histogramas alv

Segun Wikipedia:

"Un histograma es una representación gráfica de una variable en forma de barras, donde la superficie de cada barra es proporcional a la frecuencia de los valores representados. Sirven para obtener una "primera vista" general, o panorama, de la distribución de la población, o de la muestra, respecto a una característica"

Vamos a crear unos valores seudoaleatorio para graficar uno, NumPy ya cuentan con una función para generar numeros pseudoaleatorios con una distribución normal

- np.random.normal(media,des.est,muestra)

```python
#Semilla para reproducibilidad
np.random.seed(5)

#Datos
alt_mujeres = np.random.normal(157, 7, 10000)
alt_hombres = np.random.normal(170, 7, 10000)

#Gráficos
plt.figure(figsize=(5.5,4))
plt.hist(alt_mujeres,label='mujeres',bins=25,density=True,histtype='step',color='#008080')
plt.hist(alt_hombres,label='hombres',bins=25,density=True,histtype='step',color='#663399')
plt.legend(loc='upper right', fontsize=10,edgecolor='darkgray')

#Titulos y ejes
plt.title('Distribución de Alturas', fontsize=12)
plt.xlabel('Altura [cm]',fontsize=10)
plt.tick_params(axis='both', labelsize=8.5)

#Mostrar
plt.grid(True)
plt.savefig('Histograma.png',dpi=150)
plt.show()
```

El histograma nos regresa dos Arrays, uno con la cantidad de elementos que hay en cada barra y el segundo con los limites númericos de cada barra.
