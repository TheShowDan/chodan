### Limites

Y como no, empezamos con limites

$$\lim_{x \to 2} \frac{x^2+3x-10}{x^2+x-6}$$

```python
smp.limit((x**2+3*x-10)/(x**2+x-6),x,2)
```

$$\lim_{x \to e}  \tan\left(\frac{1}{\sin(x)}+ \cos(x)\right)$$

```python
smp.limit(smp.tan(1/smp.sin(x)+smp.cos(x)),x,smp.exp(1)).n()
```

$$\lim_{x \to 1^+} \frac{1}{x^2-1}$$

```python
smp.limit(1/(x**2-1),x,1,dir='+')
```

$$\lim_{x \to \infty} \frac{cos(x)-1}{x}$$

```python
smp.limit((smp.cos(x)-1) /x,x,smp.oo)
```

### Derivadas

Empezamos de chill

$$\frac{d}{dx} x^2$$

```python
smp.diff(x**2,x)
```

$$\frac{d}{dx}In\left(\frac{\sqrt{x^2+1}-x}{\sqrt{x^2-1}+x}\right)$$

```python
smp.diff(smp.log((smp.sqrt(x**2+1)-x)/(smp.sqrt(x**2-1)+x)),x)
```

$$\frac{d}{dx}tan\left(4x^4-2x^2+\frac{7}{2}x^{-3}+5\right)^{-2}$$

```python
smp.diff(smp.tan(4*x**4 - 2*x**2 + (7/2)*x**(-3) + 5)**(-2),x)
```

$$z(x)=f(x)\cdot g(x)$$

```python
f, g = smp.symbols('f, g', cls=smp.Function)
```

```python
f = f(x)
g = g(x)
z = f*g
```

```python
smp.diff(z,x)
```

### Sumas

```python
n = smp.symbols('n')
```

$$\sum_{i=1}^{5} \frac{1}{n^2}$$

```python
smp.Sum(1/n**2,(n,1,5)).doit()
```

$$\sum_{i=1}^{\infty} \frac{1}{n^2}$$

```python
smp.Sum(1/n**2,(n,1,smp.oo)).doit()
```

$$\sum_{i=2}^{\infty} \frac{2}{n^2-1}$$

```python
smp.Sum(2 / (n**2-1),(n,2,smp.oo)).doit()
```

Hay dos métodos:
- $.doit()$ - Que haga la operación
- $.n()$ - Que aproxime la operación

### Integrales Definidas

$$\int_{0}^{2} (2-x)\sqrt{x} dx$$

```python
smp.integrate((2-x)*smp.sqrt(x),(x,0,2)).n()
```

$$\int_{0}^{\frac{\pi}{2}} x cos(x^2) dx$$

```python
smp.integrate(x*smp.cos(x**2),(x,0,smp.pi/2))
```

$$\int_{0}^{t} x^7 e^x dx$$

```python
t = smp.symbols('t')
```

```python
smp.integrate(x**7 * smp.exp(x),(x,0,t))
```

___

### Integrales Indefinidas

$$\int sen^6(x) dx$$

$$\int \frac{1}{x^4+1} dx$$

```python
smp.integrate(1 / (x**4+1),x)
```
